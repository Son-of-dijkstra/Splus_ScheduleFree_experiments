```python
import matplotlib.pyplot as plt
import json
import numpy as np
```

```python
files = [
'adamw_cos_lr0.001_s0.json',    'splus_lr0.2_s2.json',
'adamw_cos_lr0.001_s1.json'  ,  'splus_noema_sf_lr0.2_s0.json',
'adamw_cos_lr0.001_s2.json'  ,  'splus_noema_sf_lr0.2_s1.json',
'adamw_sf_lr0.001_s0.json'   ,  'splus_noema_sf_lr0.2_s2.json',
'adamw_sf_lr0.001_s1.json'  ,   'splus_sf_lr0.2_s0.json',
'adamw_sf_lr0.001_s2.json' ,    'splus_sf_lr0.2_s1.json',
'splus_lr0.2_s0.json' ,         'splus_sf_lr0.2_s2.json',
'splus_lr0.2_s1.json'
]
```

```python
plt.figure(figsize=(12,8))
plt.title('Test accuracy with different setups')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
for file in files:
    with open(f"custom_experiment_results/custom_experiment_results/{file}", "r", encoding="utf-8") as f:
        data = json.load(f)
    cfg = data['cfg']
    train_loss = data['train_loss']
    test_loss = data['test_loss']
    test_acc = data['test_acc']
    time = data['time']
    if(cfg['seed']==0):
        plt.plot(np.arange(len(test_acc)), test_acc, label ='optimizer: '+cfg['method']+" seed: "+str(cfg['seed']))
    
plt.grid(True)
plt.legend()
```

```python
plt.figure(figsize=(13,10))
plt.title('Test loss with different setups')
plt.xlabel('Epoch')
plt.ylabel('Loss')
for file in files:
    with open(f"custom_experiment_results/custom_experiment_results/{file}", "r", encoding="utf-8") as f:
        data = json.load(f)
    cfg = data['cfg']
    train_loss = data['train_loss']
    test_loss = data['test_loss']
    test_acc = data['test_acc']
    time = data['time']
    if(cfg['seed']== 0):
        plt.plot(np.arange(len(test_loss)), test_loss, label ='optimizer: '+cfg['method']+" seed: "+str(cfg['seed']))
    
plt.grid(True)
plt.legend()
```
